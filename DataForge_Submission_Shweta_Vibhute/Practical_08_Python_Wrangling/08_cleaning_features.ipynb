```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv("../../01_data/raw/dataset_raw.csv")
df.head()
```

```python
print("Before:", len(df))

df = df.drop_duplicates()

print("After:", len(df))
```

```python
cat_cols = df.select_dtypes(include=["object", "string"]).columns
for c in cat_cols:
    df[c] = df[c].fillna("Unknown")
```

```python
if "Order Date" in df.columns:
    df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

    df["Order Year"] = df["Order Date"].dt.year
    df["Order Month"] = df["Order Date"].dt.month_name()

print(df.head())
```

```python
if "Sales" in df.columns:
    df["Sales Category"] = np.where(df["Sales"] >= 500, "High", "Low")

df[["Sales", "Sales Category"]].head()
```

```python
if "Sales" in df.columns:
    df["Sales Range"] = pd.cut(
        df["Sales"],
        bins=[0, 100, 500, df["Sales"].max()],
        labels=["Low", "Medium", "High"]
    )

df[["Sales", "Sales Range"]].head()
```

```python
os.makedirs("../../01_data/processed", exist_ok=True)

df.to_csv("../../01_data/processed/dataforge_cleaned.csv", index=False)

print("Data exported successfully!")
```
