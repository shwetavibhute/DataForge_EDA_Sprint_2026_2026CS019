```python
# ================================
# Practical 09 - Python EDA Tables
# ================================

import pandas as pd
import numpy as np
import os

# Load Dataset
df = pd.read_csv("../../01_data/processed/dataforge_cleaned.csv")

# Display first 5 rows
print(df.head())

# Display dataset information
print(df.info())

# Display column names
print(df.columns)

# Create output folder
os.makedirs("../outputs/tables", exist_ok=True)

# -------------------------------
# Numeric Summary
# -------------------------------
numeric_summary = df.describe(include="number").T
numeric_summary.to_csv("../outputs/tables/numeric_summary.csv")
print(numeric_summary)

# -------------------------------
# Missing Value Percentage
# -------------------------------
missing = (df.isna().mean() * 100).sort_values(ascending=False)
missing.to_csv("../outputs/tables/missing_percent.csv")
print(missing)

# -------------------------------
# Select Numeric Column
# -------------------------------
numeric_col = df.select_dtypes(include="number").columns[0]

# ===========================================
# GROUP TABLE 1 - Average Sales by Category
# ===========================================
group1 = df.groupby("Category", as_index=False)["Sales"].mean()
group1.to_csv("../outputs/tables/group_category.csv", index=False)
print(group1)

# ===========================================
# GROUP TABLE 2 - Average Sales by Region
# ===========================================
group2 = df.groupby("Region", as_index=False)["Sales"].mean()
group2.to_csv("../outputs/tables/group_region.csv", index=False)
print(group2)

# ===========================================
# GROUP TABLE 3 - Average Sales by Segment
# ===========================================
group3 = df.groupby("Segment", as_index=False)["Sales"].mean()
group3.to_csv("../outputs/tables/group_segment.csv", index=False)
print(group3)

# ===========================================
# GROUP TABLE 4 - Total Sales by Ship Mode
# ===========================================
group4 = df.groupby("Ship Mode", as_index=False)["Sales"].sum()
group4.to_csv("../outputs/tables/group_shipmode.csv", index=False)
print(group4)

# ===========================================
# TOP 10 SALES
# ===========================================
top10 = df.nlargest(10, "Sales")
top10.to_csv("../outputs/tables/top10_sales.csv", index=False)
print(top10)

# ===========================================
# BOTTOM 10 SALES
# ===========================================
bottom10 = df.nsmallest(10, "Sales")
bottom10.to_csv("../outputs/tables/bottom10_sales.csv", index=False)
print(bottom10)
bottom10.to_csv("../outputs/tables/bottom10.csv", index=False)
print(bottom10)
```

# Statistical Insights

1. The Technology category has the highest average sales among all product categories.
2. Sales differ across regions, showing that some regions generate more revenue than others.
3. Consumer, Corporate, and Home Office segments contribute differently to overall sales.
4. The Top 10 sales table highlights the highest-value orders in the dataset.
5. The Bottom 10 sales table identifies the lowest-value transactions that may require further business analysis.
